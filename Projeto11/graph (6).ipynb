```python
from google.colab import files
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
uploaded = files.upload()

for fn in uploaded.keys():
  print('"{name}" carregado com sucesso'.format(name=fn))

csv_filename = list(uploaded.keys())[0]
print(f"Arquivo '{csv_filename}' carregado")

df = pd.read_csv(csv_filename)
display(df.head())
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (14, 8)
```

```python
plt.figure()
sns.lineplot(
    data=df,
    x='frame',
    y='collisions',
    hue='strategy',
    linewidth=2.5,
    palette="viridis"
)

plt.title('Performance: Ocorrencia de colisoes', fontsize=16)
plt.xlabel('Tempo', fontsize=12)
plt.ylabel('Quantidade de colisoes simultaneas', fontsize=12)
plt.legend(title="Estrategia")
plt.show()
```

```python
plt.figure()
sns.lineplot(
    data=df,
    x='frame',
    y='quality_deviation',
    hue='strategy',
    linewidth=2.5,
    palette="magma"
)

plt.title('Qualidade: Desvio da velocidade ideal (menor é melhor)', fontsize=16)
plt.xlabel('Tempo', fontsize=12)
plt.ylabel('Erro de velocidade', fontsize=12)
plt.legend(title="Estrategia")
plt.show()
```

```python
plt.figure()
sns.lineplot(
    data=df,
    x='frame',
    y='active_agents',
    hue='strategy',
    linewidth=2.5,
    palette="coolwarm"
)

plt.title('Eficiência: Ritmo de chegada ao objetivo', fontsize=16)
plt.xlabel('Tempo', fontsize=12)
plt.ylabel('Agentes restantes no mapa', fontsize=12)
plt.legend(title="Estrategia")
plt.show()
```
